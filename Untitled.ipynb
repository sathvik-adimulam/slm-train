```python
from tokenizers import Tokenizer, models, normalizers, pre_tokenizers, Regex, decoders, trainers
```

```python
regex_pattern = r"""'(?i:[sdmt]|ll|ve|re)|
[^\r\n\p{L}\p{N}]?+\p{L}++|
\p{N}{1,3}+|
 ?[^\s\p{L}\p{N}]++[\r\n]*+|
\s++$|\s*[\r\n]|\s+(?!\S)|\s~"""
```

```python
tokenizer = Tokenizer(models.BPE())
tokenizer.normalizer = normalizers.NFC()
tokenizer.pre_tokenizer = pre_tokenizers.Sequence([
pre_tokenizers.Split(pattern=Regex(regex_pattern), behavior="isolated"),
    pre_tokenizers.ByteLevel(add_prefix_space=False, use_regex=False)
])
tokenizer.decoder = decoders.ByteLevel()
```

```python
from datasets import load_from_disk

# 1. Load your dataset from disk
# Replace 'path/to/your/dataset' with your actual directory path
dataset = load_from_disk("data/jupyter-structured-clean-dedup").take(100000)

# 2. Load a tokenizer (using GPT-2 / cl100k_base style as an example)
tokenizer = Tokenizer.from_file("tokenizer.json")


def compute_compression_ratio(batch):
  """Computes the compression ratio for a batch of texts.

  Ratio = (Length of text in bytes) / (Number of tokens) Higher ratio means
  fewer tokens represent the same amount of data.
  """
  # Tokenize the batch of text without padding/truncating for accurate counts
  encodings = tokenizer.encode_batch(batch["content"])

  ratios = []
  for text, input_ids in zip(batch["content"], encodings):
    byte_length = len(text.encode("utf-8"))
    token_count = len(input_ids.tokens)

    # Avoid division by zero for empty strings
    ratio = byte_length / token_count if token_count > 0 else 0.0
    ratios.append(ratio)

  return {"compression_ratio": ratios}


# 3. Apply the function efficiently using batched mapping
# batched=True speeds up processing significantly
processed_dataset = dataset.map(
    compute_compression_ratio,
    batched=True,
    batch_size=1000,
    remove_columns=dataset.column_names,  # Optional: keep only what you need
)

# 4. Compute the average compression ratio across the entire dataset
all_ratios = processed_dataset["compression_ratio"]
average_ratio = sum(all_ratios) / len(all_ratios)

print(f"Average Compression Ratio: {average_ratio:.4f}")
```
